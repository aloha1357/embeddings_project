## Step 4: Task 3 Part 2 - Find Word Analogies

Try to find analogies like: **woman : queen :: man : king**

**Questions:**
- Can you find analogies of this type?

```python
# Try the classic example (might not work with news corpus)
explorer.find_analogies(['woman', 'king'], ['man'], topn=10)
```

```python
# Try reverse
explorer.find_analogies(['man', 'queen'], ['woman'], topn=10)
```

```python
# Try other analogies - political/government related
analogy_attempts = [
    (['france', 'president'], ['germany']),  # country : leader
    (['berlin', 'france'], ['paris']),  # capital : country
    (['larger', 'small'], ['large']),  # comparative forms
    (['european', 'america'], ['europe']),  # adjective : noun
]

for positive, negative in analogy_attempts:
    explorer.find_analogies(positive, negative, topn=5)
```
